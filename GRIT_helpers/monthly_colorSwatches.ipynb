```python
from PIL import Image, ImageDraw, ImageFont

# ----------------------------
# GRIT Monthly Accent Swatches
# ----------------------------

months = [
    ("January",   (252, 248, 225), (210, 195, 120)),
    ("February",  (247, 233, 235), (190, 135, 145)),
    ("March",     (236, 243, 232), (150, 175, 145)),
    ("April",     (233, 240, 245), (140, 165, 185)),
    ("May",       (235, 245, 230), (135, 170, 120)),
    ("June",      (232, 242, 248), (120, 160, 190)),
    ("July",      (248, 240, 230), (200, 165, 120)),
    ("August",    (244, 238, 222), (185, 160, 105)),
    ("September", (246, 238, 215), (195, 160,  90)),
    ("October",   (247, 236, 228), (195, 135,  90)),
    ("November",  (236, 234, 230), (145, 140, 130)),
    ("December",  (234, 239, 242), (120, 145, 160)),
]

# ----------------------------
# Helpers
# ----------------------------
def rgb_to_hex(rgb):
    return "#{:02X}{:02X}{:02X}".format(*rgb)

def rgb_text(rgb):
    return f"RGB {rgb[0]}, {rgb[1]}, {rgb[2]}"

# Layout
TITLE = "GRIT Monthly Accent Color Swatches"

pad_x = 24
pad_y = 20
title_height = 54
row_height = 46

label_width = 140
swatch_width = 150
swatch_height = 28
gap = 16

# extra room for color text labels
info_width = 170

swatch_y_offset = (row_height - swatch_height) // 2

# Compute size
content_width = (
    label_width +
    gap +
    swatch_width +
    gap +
    info_width +
    gap +
    swatch_width +
    gap +
    info_width
)

img_width = pad_x + content_width + pad_x
img_height = pad_y + title_height + row_height * len(months) + pad_y

img = Image.new("RGB", (img_width, img_height), "white")
draw = ImageDraw.Draw(img)

# Fonts
try:
    title_font = ImageFont.truetype("DejaVuSans-Bold.ttf", 20)
    label_font = ImageFont.truetype("DejaVuSans.ttf", 14)
    small_font = ImageFont.truetype("DejaVuSans.ttf", 12)
except Exception:
    title_font = ImageFont.load_default()
    label_font = ImageFont.load_default()
    small_font = ImageFont.load_default()

# Title
draw.text((pad_x, pad_y), TITLE, fill="black", font=title_font)

# Column positions
x_light = pad_x + label_width + gap
x_light_info = x_light + swatch_width + 6
x_dark = x_light_info + info_width + gap
x_dark_info = x_dark + swatch_width + 6

# Optional column labels
col_label_y = pad_y + title_height - 18
draw.text((x_light, col_label_y), "Light", fill="black", font=small_font)
draw.text((x_dark,  col_label_y), "Dark",  fill="black", font=small_font)

# Rows
y0 = pad_y + title_height
for month, light, dark in months:
    y = y0

    # Month label
    draw.text((pad_x, y + (row_height - 14) // 2), month, fill="black", font=label_font)

    top = y + swatch_y_offset
    bottom = top + swatch_height

    # ---- LIGHT SWATCH ----
    draw.rectangle(
        [x_light, top, x_light + swatch_width, bottom],
        fill=light,
        outline="black",
        width=1,
    )

    # Light color text
    draw.text(
        (x_light_info, y + 6),
        f"{rgb_text(light)}\n{rgb_to_hex(light)}",
        fill="black",
        font=small_font,
    )

    # ---- DARK SWATCH ----
    draw.rectangle(
        [x_dark, top, x_dark + swatch_width, bottom],
        fill=dark,
        outline="black",
        width=1,
    )

    # Dark color text
    draw.text(
        (x_dark_info, y + 6),
        f"{rgb_text(dark)}\n{rgb_to_hex(dark)}",
        fill="black",
        font=small_font,
    )

    y0 += row_height

# Save
out_path = "monthly_accent_palette.png"
img.save(out_path, dpi=(200, 200))
print(f"Saved: {out_path} ({img_width}x{img_height})")
```
